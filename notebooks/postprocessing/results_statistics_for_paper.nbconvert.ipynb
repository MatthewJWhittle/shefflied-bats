# HSM results statistics (read-only)

This notebook **loads** project outputs and **prints/displays** tables for PDF export.  
It does **not** write CSVs, figures, or other artefacts.

**Sources:** `model_results.csv`, `tuning_results.csv`, `bats-tidy.geojson`, `aggregate_qa_latest.csv`, `data/sdm_config/**/variables_config.yml`, optional raster/shap paths.


```python
from __future__ import annotations

import json
import re
from collections import Counter
from pathlib import Path

import pandas as pd
import yaml
from IPython.display import display


def find_repo_root() -> Path:
    p = Path.cwd().resolve()
    for _ in range(8):
        if (p / "config.yml").exists():
            return p
        p = p.parent
    raise FileNotFoundError("config.yml not found; run notebook from repo or set cwd")


REPO = find_repo_root()
print("REPO =", REPO)
```

## 1. Model training summary (`model_results.csv`)


```python
mr_path = REPO / "data/sdm_models/model_results.csv"
mr = pd.read_csv(mr_path)
print(mr_path)
print("rows:", len(mr))
display(mr)
```

## 2. Hyperparameter tuning summary (`tuning_results.csv`)


```python
tr_path = REPO / "data/sdm_config/tuning_results.csv"
tr = pd.read_csv(tr_path)
print(tr_path)
display(tr)
```

## 3. Raw occurrence counts (`bats-tidy.geojson`)


```python
occ_path = REPO / "data/processed/bats-tidy.geojson"
with open(occ_path) as f:
    gj = json.load(f)
rows_occ = []
for feat in gj.get("features", []):
    p = feat.get("properties") or {}
    rows_occ.append({
        "latin_name": p.get("latin_name"),
        "activity_type": p.get("activity_type"),
    })
occ_df = pd.DataFrame(rows_occ)
occ_counts = (
    occ_df.groupby(["latin_name", "activity_type"], dropna=False)
    .size()
    .reset_index(name="raw_n")
    .sort_values(["latin_name", "activity_type"])
)
print(occ_path, "features:", len(gj.get("features", [])))
display(occ_counts)
```

## 4. Configured models vs fitted (`data/sdm_config`)


```python
cfg_dir = REPO / "data/sdm_config"
cfg_models = []
for d in sorted(cfg_dir.iterdir()):
    if not d.is_dir():
        continue
    if not (d / "variables_config.yml").exists():
        continue
    name = d.name
    if name.endswith("_Roost"):
        act, base = "Roost", name[: -len("_Roost")]
    elif name.endswith("_In_flight"):
        act, base = "In flight", name[: -len("_In_flight")]
    else:
        continue
    latin = base.replace("_", " ")
    cfg_models.append({"folder": name, "latin_name": latin, "activity_type": act})
cfg_df = pd.DataFrame(cfg_models)
fitted_keys = set(zip(mr["latin_name"], mr["activity_type"]))
cfg_df["in_model_results"] = cfg_df.apply(
    lambda r: (r["latin_name"], r["activity_type"]) in fitted_keys, axis=1
)
display(cfg_df)
missing = cfg_df[~cfg_df["in_model_results"]]
if len(missing):
    print("Configured but not in model_results.csv:")
    display(missing)
```

## 5. Aggregate QA (`aggregate_qa_latest.csv`) — p10 main


```python
qa_path = REPO / "data/sdm_predictions/aggregate_qa_latest.csv"
qa = pd.read_csv(qa_path)
print(qa_path, "rows:", len(qa))
qa_p10 = qa[qa["threshold_label"] == "p10_main"].copy()
display(qa_p10[[
    "identifier", "latin_name", "activity_type", "n_presence", "mean_cv_score",
    "threshold", "training_omission_rate", "suitable_area_percent", "warning",
]].sort_values(["activity_type", "latin_name"]))
```

## 6. Selected variables (frequency across models)


```python
freq = Counter()
per_model = []
for d in sorted(cfg_dir.iterdir()):
    if not d.is_dir():
        continue
    yp = d / "variables_config.yml"
    if not yp.exists():
        continue
    txt = yp.read_text()
    lines = []
    for line in txt.splitlines():
        if line.strip().startswith("#-"):
            continue
        if line.strip().startswith("#") and "- " in line:
            continue
        lines.append(line)
    data = yaml.safe_load("\n".join(lines)) or {}
    vars_ = [v for v in (data.get("variables") or []) if isinstance(v, str)]
    m = re.match(r"^(.+)_(Roost|In_flight)$", d.name)
    if not m:
        continue
    base, suf = m.group(1), m.group(2)
    latin = base.replace("_", " ")
    act = "Roost" if suf == "Roost" else "In flight"
    mid = f"{latin.lower().replace(' ', '_')}_{'roost' if act == 'Roost' else 'in_flight'}"
    for v in vars_:
        freq[v] += 1
    per_model.append({"model_id": mid, "n_vars": len(vars_), "variables": ", ".join(sorted(vars_))})
freq_df = pd.DataFrame(freq.most_common(), columns=["variable", "n_models"])
print("Top 25 variables by selection frequency:")
display(freq_df.head(25))
print("Per-model variable counts (wide preview):")
display(pd.DataFrame(per_model).head(20))
```

## 7. CV AUC summary (fitted models with non-null AUC)


```python
v = mr["mean_cv_score"].dropna()
summary = pd.DataFrame([
    {
        "n_models_with_auc": int(v.shape[0]),
        "min": float(v.min()),
        "max": float(v.max()),
        "median": float(v.median()),
        "mean": float(v.mean()),
        "poorest": mr.loc[v.idxmin(), "identifier"],
        "best": mr.loc[v.idxmax(), "identifier"],
    }
])
display(summary)
roost = mr[mr["activity_type"] == "Roost"]["mean_cv_score"].dropna()
flight = mr[mr["activity_type"] == "In flight"]["mean_cv_score"].dropna()
print("Roost n=", len(roost), "median AUC", roost.median())
print("In flight n=", len(flight), "median AUC", flight.median())
```
